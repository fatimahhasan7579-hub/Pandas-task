```python
#Task 1
#Create this student dataframe
import pandas as pd

# Create the DataFrame
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

average_marks = df["Marks"].mean()
maximum_marks = df["Marks"].max()
minimum_marks = df["Marks"].min()
total_students = len(df)

print(df)

print("Average marks:", average_marks)
print("Maximum marks:", maximum_marks)
print("Minimum marks:", minimum_marks)
print("Total students:", total_students)
```

```python
#Task 2
#student with marks greater than 50
print(df[df["Marks"] >= 50])

# Students with marks less than 50
print(df[df["Marks"] < 50])
```

```python
#Task 4

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Zain", "Ayesha"],
    "Course": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [70, 85, 60, 95, 45]
}

df = pd.DataFrame(data)

print(df)

average_marks = df.groupby("Course")["Marks"].mean()

print("\nAverage Marks:")
print(average_marks)
```

```python
#Task 5
import pandas as pd
import numpy as np

# DataFrame with missing marks
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [80, np.nan, 70, np.nan, 90]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Number of missing values
missing_values = df["Marks"].isnull().sum()
print("\nNumber of missing values:", missing_values)

# Average marks
average_marks = df["Marks"].mean()

# Fill missing values with average
df["Marks"] = df["Marks"].fillna(average_marks)

print("\nCleaned DataFrame:")
print(df)
```

```python
#14. Handling Missing Values
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

print(df)

print(df.isnull())

print(df.isnull().sum())

df = df.dropna()
print(df)

df["Marks"] = df["Marks"].fillna(0)
print(df)
```

```python
#15. Handling Duplicate Data
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ali"],
    "Marks": [80, 90, 80]
}

df = pd.DataFrame(data)

print(df.duplicated())

print(df.duplicated().sum())

df = df.drop_duplicates()
print(df)
```

```python
#17. Aggregation and GroupBy
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [70, 95, 80]
}

df = pd.DataFrame(data)

marks = df["Marks"]

print(marks.sum())
print(marks.mean())
print(marks.max())
print(marks.min())
print(marks.count())
```

```python
#18. Combining DataFrames
import pandas as pd

df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Marks": [75, 85]
})

final_df = pd.concat([df1, df2], ignore_index=True)

print(final_df)
```

```python
#merge()
import pandas as pd

students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})

final_df = pd.merge(students, marks, on="ID")

print(final_df)
```
